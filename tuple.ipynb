```python
# Packing
Students = ("Maha", 23, "Data Analyst")

# Unpacking
Name, Age, Course = Students

print(Name)
print(Age)
print(Course)

Age = Age + 1
Students = (Name, Age, Course)

print(Students)

```

```python
# Packing
tuple = (105, "Laptop", 15, 25000, "Delivered")

# Unpacking
order_id, product_name, quantity, price, status = tuple

print(tuple)
print(f"Order Id: {order_id}")
print(f"Product Name: {product_name}")
print(f"Quantity: {quantity}")
print(f"Price: {price}")
print(f"Staturs: {status}")




```
