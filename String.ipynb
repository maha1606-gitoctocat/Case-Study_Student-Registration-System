```python
# String Concatenation
String1 = "Hello "

# Get user input
Name = input("Enter your Name: ")
Result = String1 + Name

# Display output
print(Result)

String3 = ", Welcome to Python programming"
Final_Result = Result + String3

# Display final output
print(f"Final Result: {Final_Result}")
```

```python
# String Slicing and Indexing
Text = "Hello Zara, welcome to Python programming"

First_Character = Text[0]
Last_Character = Text[-1]
First_5_Characters = Text[:5]
Last_11_Characters = Text[-11]
Reversed_String = Text[-11:]
Word_Using_Slicing = Text[23:29]

#  Display Details
print(f"First character: {First_Character}")
print(f"Last character: {Last_Character}")
print(f"First 5 characters: {First_5_Characters}")
print(f"Last 11 characters: {Last_11_Characters}")
print(f"Reversed string: {Reversed_String}")
print(f"Word using slicing: {Word_Using_Slicing}")

```

```python
# String Methods
strM = "Python beginner tutorial"

strM_Upper = (strM).upper()
strM_Lower = (strM).lower()
strM_Capitalize = (strM).capitalize()
strM_Count = (strM).count("t")
strM_Replace = (strM).replace("Python", "Data Analytics")

# Display Details

print(f"StrM_Upper: {strM_Upper}")
print(f"strM_Lower: {strM_Lower}")
print(f"strM_Capitalize: {strM_Capitalize}")
print(f"strM_Count: {strM_Count}")
print(f"strM_Replace: {strM_Replace}")

```
