```python
# Initial List
Name = ["Maha", "Lavanya","Dina", "Selvi", "Akash"]

# Update name
Name[2] = "Akash"

print(Name)

# append
Name.append("Royal")
print(Name)

# extend
Name.extend(["Ravi", "Kumar"])
print(Name)

# Insert
Name.insert(2, "Anusha")
print (Name)

# Remove
Name.remove("Ravi")
print(Name)

# pop
Name.pop()
print(Name)

# Pop index
Name.pop(3)
print(Name)

# sort
Name.sort()
print(Name)

# sorted
print(sorted(Name))

# reverse
Name.reverse()
print(Name)
```

```python
students_marks = list(map(int,input("Enter the students marks: ").split(",")))
print(students_marks)

Total_marks = sum(students_marks)
print(f"Total Marks: {Total_marks}")

Average = Total_marks/len(students_marks)
print(f"Average: {Average}")

Percentage = (Total_marks/500)*100
print(f"Percentage: {Percentage}")


```
