```python
# Input details
Full_Name = input("Enter the Name : ").lower().replace(" ", "_")
Domain_Name = input("Enter the Domain: ")

# Create Email
Email = Full_Name + "@" + Domain_Name

# Display details
print(f"Full Name: {Full_Name}")
print(f"Domain Name: {Domain_Name}")
print(f"Email: {Email}")
```
