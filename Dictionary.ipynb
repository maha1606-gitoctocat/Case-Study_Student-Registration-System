```python
# Case study - Student Data Management System

student = {"Name": ["Maha", "Akash", "Lavanya"],
           "Age" : [23, 23, 20],
           "Course" : ["Data Analyst", "Java", "Python"]
           }
# Add new student
student["Name"].append("Dinakaran")
student["Age"].append(20)
student["Course"].append("Data Engineering")


# Print Datas
print(f"All Names: {student['Name']}")
print(f"All Ages: {student['Age']}")
print(f"All Courses: {student['Course']}")

# Display First Student details
print("\n----- Second Student -----")
print(f"Name: {student['Name'][1]}")
print(f"Age: {student['Age'][1]}")
print(f"Course: {student['Course'][1]}")

# Display Newly Added Student
print("\n----- Newly Added Student -----")
print(f"Name: {student['Name'][3]}")
print(f"Age: {student['Age'][3]}")
print(f"Course: {student['Course'][3]}")

```
