```python
# **Case Study – Shopping Cart System**
**Instructions:**
1. Create a list named `cart` with initial items:
   Milk, Bread, Eggs, Apple
2. Perform list operations:
 • Add multiple items using extend()
 • Insert item at index 2 using insert()
 • Remove item Apple using remove()
 • Remove last item using pop()
3. Use list methods:
   • index() to find position of Eggs
   • count() to count occurrences of Milk
4. Apply indexing:
   • Display first item
   • Display last item
5. Apply slicing:
   • Display first 3 items from the list
6. Find total number of items using len()
7. Display all outputs using print()
```

```python
# **Case Study – Shopping Cart System**
Cart = ["Milk", "Bread", "Eggs", "Apple"]
print(Cart)

# extend
Cart.extend(["Butter", "Curd", "Jam"])
print(Cart)

# Insert
Cart.insert(2, "Cheese")
print(Cart)

# Remove
Cart.remove("Apple")
print(Cart)

# Position of egg
Cart[2]

# Count of Milk
Cart.count("Milk")

# Display first item
Cart[0]

# Display last item
Cart[3]

# First 3 Items
Cart[0:3]

# len
len(Cart)

# Display Outputs
print(f"Cart: {Cart}")
print(f"First Item: {Cart[0]}")
print(f"Last Item: {Cart[3]}")
print(f"First 3 Items: {Cart[0:3]}")
print(f"Total Items: {len(Cart)}")



```
