```python
# Tuples
First_tuple = (10, 20, 30)
Second_tuple = (40, 50, 60)

t_combine = First_tuple + Second_tuple
Repeat_Elements = t_combine * 3
Third_Element = t_combine[2]
First_3_Elements = t_combine[:3]
Last_3_Elements = t_combine[-3:]

# Display Outputs
print(f"t_combine: {t_combine}")
print(f"Repeat_Elements: {Repeat_Elements}")
print(f"Third_Element: {Third_Element}")
print(f"First_3_Elements: {First_3_Elements}")
print(f"Last_3_Elements: {Last_3_Elements}")
```
